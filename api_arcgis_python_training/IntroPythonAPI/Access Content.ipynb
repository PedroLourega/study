# Retrieve and review portal items

## Prepare workspace

```python
from arcgis.gis import GIS
```

```python
gis = GIS()
```

```python
gis
```

## Access a portal item
Retrieve a Feature Layer hosted on another organization.

### Access using search()

```python
search_results = gis.content.search("traffic incidents AND owner:EsriTrainingSvc")
```

```python
search_results
```

```python
search_results2 = gis.content.search("traffic incidents AND owner:EsriTrainingSvc", item_type="Feature*", outside_org=True)
```

```python
search_results2
```

### Access using get()

```python
incidents = search_results[5]
```

```python
incidents
```

```python
incidents_test_get = gis.content.get('320838aca5aa4d53918f3c92d3549975')
```

```python
incidents_test_get
```
